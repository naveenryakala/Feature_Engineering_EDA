# Data Cleaning - Google Playstore Dataset

In this notebook, we will clean the Google Playstore dataset by performing the following steps:
- Handling missing values
- Removing duplicates
- Correcting data types
- Removing invalid entries

```python
# Import necessary libraries
import pandas as pd
import numpy as np

# Display settings for better visualization
pd.set_option('display.max_columns', None)
```

## Load the Dataset
We will load the Google Playstore dataset into a Pandas DataFrame.

```python
# Load the dataset
file_path = '../datasets/google_playstore.csv'  # Adjust the path as needed
data = pd.read_csv(file_path)

# Display the first few rows of the dataset
data.head()
```

## Step 1: Handle Missing Values
We will check for missing values and handle them appropriately.

```python
# Check for missing values
print("Missing values per column:")
print(data.isnull().sum())

# Drop rows with missing values (if necessary)
data = data.dropna()

# Verify that there are no missing values left
print("Missing values after cleaning:")
print(data.isnull().sum())
```

## Step 2: Remove Duplicates
We will check for duplicate rows and remove them.

```python
# Check for duplicate rows
print(f"Number of duplicate rows: {data.duplicated().sum()}")

# Remove duplicate rows
data = data.drop_duplicates()

# Verify that duplicates are removed
print(f"Number of duplicate rows after cleaning: {data.duplicated().sum()}")
```

## Step 3: Correct Data Types
We will ensure that all columns have the correct data types.

```python
# Convert 'Reviews' column to numeric
data['Reviews'] = pd.to_numeric(data['Reviews'], errors='coerce')

# Remove '$' from 'Price' column and convert to numeric
data['Price'] = data['Price'].str.replace('$', '').astype(float)

# Remove 'M' from 'Size' column and convert to numeric (assuming size is in MB)
data['Size'] = data['Size'].str.replace('M', '').replace('Varies with device', np.nan).astype(float)

# Verify data types
print(data.dtypes)
```

## Step 4: Remove Invalid Entries
We will remove rows with invalid or inconsistent data.

```python
# Remove rows where 'Rating' is greater than 5 (invalid ratings)
data = data[data['Rating'] <= 5]

# Verify the dataset after cleaning
data.describe()
```

## Save the Cleaned Dataset
We will save the cleaned dataset to a new CSV file.

```python
# Save the cleaned dataset
cleaned_file_path = '../datasets/google_playstore_cleaned.csv'  # Adjust the path as needed
data.to_csv(cleaned_file_path, index=False)
print(f"Cleaned dataset saved to {cleaned_file_path}")
```
