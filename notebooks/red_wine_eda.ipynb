# Exploratory Data Analysis on Red Wine Dataset

In this notebook, we will perform exploratory data analysis (EDA) on the Red Wine dataset. We will visualize and summarize key insights to understand the dataset better.

```python
# Importing necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Setting the aesthetic style of the plots
sns.set(style='whitegrid')
```

```python
# Loading the dataset
red_wine_data = pd.read_csv('../datasets/red_wine.csv')

# Displaying the first few rows of the dataset
red_wine_data.head()
```

```python
# Checking the shape of the dataset
red_wine_data.shape
```

```python
# Checking for missing values
missing_values = red_wine_data.isnull().sum()
missing_values[missing_values > 0]
```

```python
# Descriptive statistics of the dataset
red_wine_data.describe()
```

```python
# Visualizing the distribution of the target variable (quality)
plt.figure(figsize=(10, 6))
sns.countplot(x='quality', data=red_wine_data)
plt.title('Distribution of Wine Quality')
plt.xlabel('Quality')
plt.ylabel('Count')
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(12, 8))
correlation = red_wine_data.corr()
sns.heatmap(correlation, annot=True, fmt='.2f', cmap='coolwarm', square=True)
plt.title('Correlation Heatmap')
plt.show()
```

```python
# Pairplot to visualize relationships between features
sns.pairplot(red_wine_data, hue='quality')
plt.show()
```

## Conclusion

In this notebook, we performed exploratory data analysis on the Red Wine dataset. We visualized the distribution of the target variable, checked for missing values, and explored the relationships between features. Further analysis and feature engineering can be conducted based on these insights.
