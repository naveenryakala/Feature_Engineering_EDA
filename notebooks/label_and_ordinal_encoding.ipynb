# Label and Ordinal Encoding

This notebook covers the techniques of label encoding and ordinal encoding for categorical data. These methods are essential for preparing categorical variables for machine learning algorithms.

```python
# Import necessary libraries
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder

# Load dataset
data = pd.read_csv('../datasets/red_wine.csv')
data.head()
```

## Label Encoding

Label encoding converts categorical labels into numerical values. This is useful for algorithms that can only work with numerical input.


```python
# Initialize LabelEncoder
label_encoder = LabelEncoder()

# Example of label encoding
data['quality_encoded'] = label_encoder.fit_transform(data['quality'])
data[['quality', 'quality_encoded']].head()
```

## Ordinal Encoding

Ordinal encoding is used when the categorical variable has a meaningful order. This method assigns integers to categories based on their order.


```python
# Map quality to ordinal categories
quality_mapping = {5: 'low', 6: 'medium', 7: 'high'}
data['quality_category'] = data['quality'].map(quality_mapping)

# Initialize OrdinalEncoder
ordinal_encoder = OrdinalEncoder(categories=[['low', 'medium', 'high']])

# Example of ordinal encoding
data['quality_ordinal'] = ordinal_encoder.fit_transform(data[['quality_category']])
data[['quality', 'quality_category', 'quality_ordinal']].head()
```

## Conclusion

In this notebook, we demonstrated how to perform label encoding and ordinal encoding on categorical data. These techniques are crucial for preparing data for machine learning models.
