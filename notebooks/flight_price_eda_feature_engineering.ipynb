# Flight Price EDA and Feature Engineering

In this notebook, we will perform exploratory data analysis (EDA) and feature engineering on the flight price dataset. We will analyze the dataset to extract insights and prepare it for machine learning models.

```python
# Importing necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set the visualisation style
sns.set(style='whitegrid')
```

```python
# Load the dataset
flight_data = pd.read_csv('../datasets/flight_price.csv')

# Display the first few rows of the dataset
flight_data.head()
```

```python
# Check for missing values
missing_values = flight_data.isnull().sum()
missing_values[missing_values > 0]
```

```python
# Handling missing values
# For simplicity, we will drop rows with missing values in this example
flight_data_cleaned = flight_data.dropna()

# Check the shape of the cleaned dataset
flight_data_cleaned.shape
```

```python
# Exploratory Data Analysis
# Visualizing the distribution of flight prices
plt.figure(figsize=(10, 6))
sns.histplot(flight_data_cleaned['Price'], bins=30, kde=True)
plt.title('Distribution of Flight Prices')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()
```

```python
# Feature Engineering
# Creating new features based on existing ones
flight_data_cleaned['Journey_Duration'] = flight_data_cleaned['Arrival_Time'] - flight_data_cleaned['Departure_Time']

# Display the first few rows of the updated dataset
flight_data_cleaned.head()
```

```python
# Encoding categorical variables
flight_data_encoded = pd.get_dummies(flight_data_cleaned, columns=['Airline', 'Source', 'Destination'], drop_first=True)

# Display the first few rows of the encoded dataset
flight_data_encoded.head()
```

```python
# Final dataset ready for modeling
final_dataset = flight_data_encoded.drop(['Arrival_Time', 'Departure_Time'], axis=1)

# Display the shape of the final dataset
final_dataset.shape
```

## Conclusion

In this notebook, we performed exploratory data analysis and feature engineering on the flight price dataset. We handled missing values, visualized the distribution of flight prices, created new features, and encoded categorical variables to prepare the dataset for machine learning.
