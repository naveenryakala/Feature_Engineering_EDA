# Handling Imbalanced Dataset

In this notebook, we will explore various methods to handle imbalanced datasets. Imbalanced datasets can lead to biased models that perform poorly on the minority class. We will discuss several techniques to address this issue.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
from imblearn.over_sampling import SMOTE

# Set random seed for reproducibility
np.random.seed(42)
```

## Load Dataset
We will use the flight price dataset for this demonstration. Let's load the dataset and check its structure.

```python
# Load the dataset
data = pd.read_csv('../datasets/flight_price.csv')
data.head()
```

## Check for Class Imbalance
Let's check the distribution of the target variable to see if there is an imbalance.

```python
# Check the distribution of the Airline column
data['Airline'].value_counts()
```

## Splitting the Data
We will split the data into training and testing sets.

```python
# Split the data into features and target
X = data.drop('Price', axis=1)
y = data['Price']

# Ensure each class in y has at least two members
class_counts = y.value_counts()
valid_classes = class_counts[class_counts > 1].index
data_filtered = data[data['Price'].isin(valid_classes)]

# Update X and y after filtering
X = data_filtered.drop('Price', axis=1)
y = data_filtered['Price']

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=42, stratify=y)
```

## Handling Imbalance with SMOTE
We will use the SMOTE technique to handle the class imbalance in the training dataset.

```python
# Apply SMOTE to the training data
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X_train, y_train)

# Check the new class distribution
pd.Series(y_resampled).value_counts()
```

## Model Training
Now, we will train a Random Forest classifier on the resampled dataset.

```python
# Ensure X_resampled and y_resampled are defined
if 'X_resampled' not in locals() or 'y_resampled' not in locals():
	# Apply SMOTE to the training data as a fallback
	smote = SMOTE(random_state=42)
	X_resampled, y_resampled = smote.fit_resample(X_train, y_train)

# Train a Random Forest classifier
model = RandomForestClassifier(random_state=42)
model.fit(X_resampled, y_resampled)
```

## Model Evaluation
Finally, we will evaluate the model on the test set.

```python
# Ensure the model is trained before evaluation
if not hasattr(model, "estimators_"):
	# Train the model if not already trained
	smote = SMOTE(random_state=42)
	X_resampled, y_resampled = smote.fit_resample(X_train, y_train)
	model.fit(X_resampled, y_resampled)

# Evaluate the model
y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```
