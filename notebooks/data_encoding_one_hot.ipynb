# One-Hot Encoding

This notebook demonstrates how to perform one-hot encoding for nominal data. One-hot encoding is a method to convert categorical variables into a format that can be provided to machine learning algorithms to improve predictions.

```python
# Importing necessary libraries
import pandas as pd

# Load the dataset
data = pd.read_csv('../datasets/red_wine.csv')
data.head()
```

## Understanding the Dataset
Let's take a look at the dataset to understand its structure and the categorical variables that need to be encoded.

```python
# Displaying the dataset information
data.info()
```

## One-Hot Encoding
We will perform one-hot encoding on the categorical variables in the dataset. This will create new binary columns for each category.

```python
# Performing one-hot encoding
data_encoded = pd.get_dummies(data, drop_first=True)
data_encoded.head()
```

## Conclusion
In this notebook, we demonstrated how to perform one-hot encoding on a dataset. This technique is essential for preparing categorical data for machine learning algorithms.
