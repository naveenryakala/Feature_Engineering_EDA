# Exploratory Data Analysis on Cleaned Google Playstore Dataset

In this notebook, we will perform exploratory data analysis (EDA) on the cleaned Google Playstore dataset. The goal is to visualize and summarize key insights from the dataset.

```python
# Importing necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Setting the aesthetic style of the plots
sns.set_style('whitegrid')
```

```python
# Loading the cleaned Google Playstore dataset
df = pd.read_csv('../datasets/google_playstore_cleaned.csv')

# Displaying the first few rows of the dataset
df.head()
```

```python
# Checking the shape of the dataset
df.shape
```

```python
# Checking for missing values
missing_values = df.isnull().sum()
missing_values[missing_values > 0]
```

```python
# Visualizing the distribution of app ratings
plt.figure(figsize=(10, 6))
sns.histplot(df['Rating'], bins=30, kde=True)
plt.title('Distribution of App Ratings')
plt.xlabel('Rating')
plt.ylabel('Frequency')
plt.show()
```

```python
# Analyzing the relationship between rating and number of installs
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Installs', y='Rating', data=df)
plt.title('Rating vs. Number of Installs')
plt.xlabel('Number of Installs')
plt.ylabel('Rating')
plt.xscale('log')
plt.show()
```

```python
# Analyzing the distribution of app categories
plt.figure(figsize=(12, 8))
sns.countplot(y='Category', data=df, order=df['Category'].value_counts().index)
plt.title('Number of Apps in Each Category')
plt.xlabel('Number of Apps')
plt.ylabel('Category')
plt.show()
```

```python
# Analyzing the correlation between numerical features
plt.figure(figsize=(10, 8))
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

## Conclusion

In this notebook, we performed exploratory data analysis on the cleaned Google Playstore dataset. We visualized the distribution of app ratings, analyzed the relationship between ratings and installs, and examined the distribution of app categories. Additionally, we explored the correlation between numerical features.
