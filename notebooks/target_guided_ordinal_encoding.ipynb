# Target Guided Ordinal Encoding

In this notebook, we will explore the concept of target guided ordinal encoding. This technique is useful for encoding ordinal variables based on the relationship with the target variable.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualisation style
sns.set(style='whitegrid')
```

## Target Guided Ordinal Encoding
We will implement target guided ordinal encoding for the ordinal features in our dataset. This involves calculating the mean of the target variable for each category of the ordinal feature.

```python
# Function for target guided ordinal encoding
def target_guided_encoding(data, feature, target):
    # Calculate the mean of the target for each category
    mean_target = data.groupby(feature)[target].mean().to_dict()
    # Map the mean target to the feature
    data[feature + '_encoded'] = data[feature].map(mean_target)
    return data

# Example usage
data = target_guided_encoding(data, 'quality', 'quality')  # Using 'quality' as the target column
data[['quality', 'quality_encoded']].head()
```
